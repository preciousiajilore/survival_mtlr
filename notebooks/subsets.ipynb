```python
def split_lengths(val):
    if pd.isna(val):
        return (0.0, 0.0)
    s = str(val)
    # Find all numbers with optional decimal/comma
    nums = re.findall(r'[\d]+[.,]?[\d]*', s)
    nums = [n.replace(',', '.') for n in nums]  # replace comma with dot
    nums = [float(n) for n in nums]
    # If only one length, fill second column with 0.0
    if len(nums) == 1:
        return (nums[0], 0.0)
    elif len(nums) >= 2:
        return (nums[0], nums[1])
    else:
        return (0.0, 0.0)
```

```python
#Apply it
df[['stxlength_1', 'stxlength_2']] = df['stxlength'].apply(split_lengths).apply(pd.Series)
```

```python
def clean_spcomp(val):
    if pd.isna(val):
        return np.nan
    v = str(val).strip().lower()
    # Handle "no" values
    if v in ['no', 'n', '0', 'n/a']:
        return 0
    # Handle "yes" values, or if the text contains "yes"
    if v.startswith('yes') or v == '1' or 'yes' in v:
        return 1
    # If it’s just "fell out" or other text, count as complication (1)
    if 'fell out' in v or 'infection' in v or 'fistula' in v or 'drainage' in v or 'occluded' in v or 'clots' in v or 'irrigation' in v or 'encrusted' in v:
        return 1
    # Anything else, set as nan
    return np.nan

df["spcomplication_clean"] = df["spcomplication"].apply(clean_spcomp)
```

```python
import numpy as np

def clean_erectilepre(val):
    if pd.isna(val):
        return np.nan
    v = str(val).lower().strip()
    # 1. Gender reassignment gets 0
    if 'gender reassignment' in v:
        return 0
    # 2. Obvious yes answers (any variant of "yes")
    if v.startswith('yes') or v.startswith('y;') or v == '1' or 'yes' in v:
        return 1
    # 3. Obvious no answers
    if v.startswith('no') or v == '0' or 'no' in v or 'not sexually active' in v:
        return 0
    # 4. Not sexually active or unclear? Make missing for now.
    if '?' in v or 'probably' in v or 'no records' in v:
        return np.nan
    # 5. If it contains a number in the yes (e.g., yes (5/10)), treat as 1
    if '/' in v and 'yes' in v:
        return 1
    # 6. Default to nan if not caught above
    return np.nan

df["erectilepre_clean"] = df["erectilepre"].apply(clean_erectilepre)
```

```python
def clean_cathdays(val):
    # Handle missing/empty
    if pd.isna(val):
        return np.nan
    s = str(val).lower().strip()
    # If contains "no record", "not recorded", "unclear", etc., return NaN
    if ('no record' in s or 'not recorded' in s or 'unclear' in s or 
        'still in' in s or 'not removed' in s or '?' in s or
        'follw up not in chart' in s or 'follow-up not in chart' in s or
        'unknown' in s):
        return np.nan
    # Try to extract the first number found (including negatives, e.g., -44501)
    nums = re.findall(r'-?\d+', s)
    if nums:
        return float(nums[0])
    return np.nan

df['cathdays_clean'] = df['cathdays'].apply(clean_cathdays)
```

```python
import numpy as np

def clean_erectilepost(val):
    if pd.isna(val):
        return np.nan
    v = str(val).lower().strip()
    # Handle pure numeric codes
    if v in ['0', '1', '2', '3', '4']:
        return int(v)
    # Handle "no" (with or without text)
    if v == 'no' or v == 'no comments on function' or v == 'pre-op dysfunction':
        return 0
    # Handle special cases that contain "1"
    if v.startswith('1') or 'same' in v or 'cialis' in v or 'no comments' in v:
        return 1
    # Handle special cases that contain "2"
    if v.startswith('2') or 'worse' in v or 'significant ed' in v:
        return 2
    # Handle "3" or "ejac" for ejaculatory dysfunction
    if v.startswith('3') or 'ejac' in v:
        return 3
    # If "not mentioned" or "gender reassignment", treat as missing
    if 'not mentioned' in v or 'gender reassign' in v:
        return np.nan
    # If nothing matches, leave as NaN
    return np.nan

df["erectilepost_clean"] = df["erectilepost"].apply(clean_erectilepost)
```
