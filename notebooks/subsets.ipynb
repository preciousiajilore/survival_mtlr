The purpose of this code is to create the four subsets of the data.
They are as follows:
    1) Before Surgery
    2) Immediately Before Surgery
    3) Immediately after Surgery
    4) Post Surgery

Additionally, we have the outcome variable which is the target variable (there are a couple that we have but we will use Failure for now)


```python
#Imports:
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import re 

```

```python
#Load the dictionary
dictionary = pd.read_csv("/Users/preciousajilore/Documents/GitHub/torchmtlr/data/featuredictionary_new.csv")

dictionary.columns = [col.strip() for col in dictionary.columns]
```

```python
dictionary.columns
```

```python
#Get a list of features considered as "before surgery"
before_surgery = dictionary.loc[dictionary['feature category'].str.strip().str.lower() == 'before surgery', 'abbreviated name'].tolist()
```

```python
print(before_surgery)
```

```python
#Get a list of features considered to be "immediately before surgery"
imm_bf_surgery = dictionary.loc[dictionary['feature category'].str.strip().str.lower() == 'immediately before surgery', 'abbreviated name'].tolist()

print(imm_bf_surgery)
```

```python
#Get a list of features considered to be "immediately after surgery"
imm_aft_surgery = dictionary.loc[dictionary['feature category'].str.strip().str.lower() == 'immediately after surgery',"abbreviated name"].tolist()

print(imm_aft_surgery)
```

```python
#get list of features considered to be "post-surgery"
post_surgery = dictionary.loc[dictionary['feature category'].str.strip().str.lower() == 'post surgery',"abbreviated name"].tolist()

print(post_surgery)
```

```python
#get list for target variables
outcome_vars = dictionary.loc[dictionary['feature category'].str.strip().str.lower() == 'outcome to predict',"abbreviated name"].tolist()

print(outcome_vars)
```

```python
before_surgery = [f.lower().strip().replace(" ","") for f in before_surgery]
imm_bf_surgery = [f.lower().strip().replace(" ","") for f in imm_bf_surgery]
imm_aft_surgery = [f.lower().strip().replace(" ", "") for f in imm_aft_surgery]
post_surgery = [f.lower().strip().replace(" ","") for f in post_surgery]
```

```python
print(imm_aft_surgery)
```

```python
excelSheet = pd.ExcelFile('/Users/preciousajilore/Documents/GitHub/torchmtlr/notebooks/first_analysis.xlsx')
print(excelSheet.sheet_names)
```

```python
df = pd.read_excel(
    excelSheet,
    sheet_name= 'Sheet1',
    engine = 'openpyxl'
)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ","")
```

```python
df.columns
```

```python
#Making the subsets
x_before = df[before_surgery]
x_imm_bf = df[imm_bf_surgery]
x_imm_af = df[imm_aft_surgery]
x_post = df[post_surgery]
```

```python
outcomes = df[outcome_vars]

x_before_with_outcomes = x_before.join(outcomes)
x_imm_bf = x_imm_bf.join(outcomes)
x_imm_af = x_imm_af.join(outcomes)
x_post = x_post.join(outcomes)
```
